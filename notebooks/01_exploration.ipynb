```python
import warnings; warnings.filterwarnings("ignore")
import pandas as pd
from statsbombpy import sb

matchs = sb.matches(competition_id=11, season_id=27)   # 11 = La Liga, 27 = 2015/16
barca = matchs[(matchs.home_team == "Barcelona") | (matchs.away_team == "Barcelona")]
print(len(barca), "matchs du Barça")
barca[["match_id", "match_date", "home_team", "away_team", "home_score", "away_score"]].sort_values("match_date").head()
```

```python
match_id = barca.sort_values("match_date").match_id.iloc[0]
ev = sb.events(match_id=match_id)
print(ev.shape)
ev.type.value_counts().head(10)
```

```python
ev[["minute", "team", "player", "type", "location", "pass_end_location", "pass_outcome", "possession"]].head(20)
```

```python
from mplsoccer import Pitch

passes = ev[(ev.team == "Barcelona") & (ev.type == "Pass")]
reussies = passes[passes.pass_outcome.isna()]
print(f"{len(passes)} passes, {len(reussies)} réussies ({len(reussies)/len(passes):.0%})")

pitch = Pitch(pitch_type="statsbomb")
fig, ax = pitch.draw(figsize=(10, 7))
pitch.arrows(reussies.location.str[0], reussies.location.str[1],
             reussies.pass_end_location.str[0], reussies.pass_end_location.str[1],
             ax=ax, width=1, alpha=0.4, color="#a50044")
ax.set_title("Passes réussies du Barça – 1er match de Liga 2015/16")
fig.savefig("../figures/passes_match1.png", dpi=120, bbox_inches="tight")
```
