```python
import sys
sys.path.append("../src")          # pour que Python trouve possessions.py
import pandas as pd
from possessions import charger_saison, actions_barca

ev = charger_saison(season_id=27, dossier="../data")   # lit le cache, instantané
a = actions_barca(ev)
seq = pd.read_pickle("../data/sequences_1516.pkl")
seq.head(10)
```

```python
from mplsoccer import Pitch

ex = seq[seq.fin == "but"].iloc[0]
g = a[(a.match_id == ex.match_id) & (a.possession == ex.possession)]
print(g[["minute", "player", "type", "x", "y", "zone"]])

pitch = Pitch(pitch_type="statsbomb")
fig, ax = pitch.draw(figsize=(10, 7))
pitch.lines(g.x[:-1], g.y[:-1], g.x[1:], g.y[1:], ax=ax, color="#a50044", lw=2, comet=True)
pitch.scatter(g.x, g.y, ax=ax, s=80, color="#004d98", zorder=3)
for k, (x, y) in enumerate(zip(g.x, g.y), 1):
    ax.annotate(str(k), (x, y), xytext=(4, 4), textcoords="offset points")
ax.set_title("Une possession du Barça qui finit en but")
fig.savefig("../figures/possession_but.png", dpi=110, bbox_inches="tight")
```

```python
part_but = (seq.fin == "but").mean()
part_tir = seq.tir.mean()

print(f"Possessions qui finissent en but : {part_but:.1%}")
print(f"Possessions avec au moins un tir : {part_tir:.1%}")
```

```python
n_buts = (seq.fin == "but").sum()
n_avec_tir = seq.tir.sum()

print(f"{n_buts} buts sur {n_avec_tir} possessions avec un tir")
print(f"Taux de conversion : {n_buts / n_avec_tir:.1%}")
```

```python
print(seq.n_actions.describe().round(1))

n_une_action = (seq.n_actions == 1).sum()
print(f"Possessions d'une seule action : {n_une_action}")
```

```python
long_buts = seq[seq.fin == "but"].n_actions.mean()
long_toutes = seq.n_actions.mean()

print(f"Longueur moyenne, possessions avec but : {long_buts:.1f}")
print(f"Longueur moyenne, toutes possessions : {long_toutes:.1f}")
```
